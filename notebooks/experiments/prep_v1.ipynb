# Prep v1

Prepare clean EURUSD H1 data for feature engineering.

```python
import pandas as pd
import numpy as np
```

```python
# file path
path = r"C:\Users\LENOVO\AlgoTrading_Research\statistical-range-detection\data\raw\EURUSD_H1_2022_2025_raw.csv"

# load dataset & column names
df = pd.read_csv(path, sep=",", header=None, encoding="utf-16", names=["date", "open", "high", "low", "close", "tick_volume", "spread"], parse_dates=["date"], index_col="date")

df.head()
```

```python
df = df.sort_index()

df.head()
```

```python
df.index.is_monotonic_increasing
```

```python
df.index.duplicated().sum()
```

```python
df["returns"] = df["close"].pct_change()

df.head()
```

```python
df["log_returns"] = np.log(df["close"] / df["close"].shift(1))

df.head()
```

```python
df.isnull().sum()
```

```python
df = df.dropna()

df.head()
```

```python
df.isnull().sum()
```
