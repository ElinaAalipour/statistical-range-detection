```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

file_path = r"C:\Users\LENOVO\AlgoTrading_Research\statistical-range-detection\data\raw\EURUSD_H1_2022_2025_raw.csv"
df = pd.read_csv(file_path, sep=",", header=None, encoding="utf-16")
df.columns = ["datetime", "open", "high", "low", "close", "tick_volume", "spread"]

df["datetime"] = pd.to_datetime(df["datetime"])

df.head()
```

```python
df.set_index("datetime", inplace=True)
```

```python
plt.figure(figsize=(12,8))
plt.plot(df["datetime"], df["close"])

plt.title("EURUSD H1 Price")
plt.xlabel("Date")
plt.ylabel("Price")

plt.show()
```
